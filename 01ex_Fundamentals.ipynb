3\. Write a `decorator` hello that makes every wrapped function print “Hello!”, i.e. something like:

```python
@hello
def square(x):
    return x*x
```



````python
from functools import wraps
from collections.abc import Callable
from typing import Any

def hello(target_function: Callable):
    @wraps(target_function)
    def wrapper(*args, **kwargs) -> Any:
        print("Hello!")
        return target_function(*args, **kwargs)
    
    return wrapper

@hello
def square(x: float) -> float:
    return x*x

square()
````

4\. Write the factorial function so that it a) does and b) does not use recursion.

````python
from functools import reduce, lru_cache


@lru_cache(maxsize=256)
def fac_rec(n: int) -> int:
    if n < 1: raise ValueError("Invalid fac number")
    if n == 1: return 1
    return n * fac_rec(n-1)


def fac_non_rec(n: int) -> int:
    result = reduce(lambda x,y: x* y, [m for m in range(n)])


````

5. Use HOFs (zip in particular) to compute the weight of a circle, a disk and a sphere, assuming different radii and different densities:

```python
densities = {"Al":[0.5,1,2],"Fe":[3,4,5],"Pb": [15,20,30]}
radii = [1,2,3]
```

where the entries of the dictionary's values are the linear, superficial and volumetric densities of the materials respectively.

In particular define a list of three lambda functions using a comprehension that computes the circumference, the area and the volume for a given radius.


````python

from math import pi
from typing import NamedTuple

DensityTuple = NamedTuple("DensityTuple", [('rho_lin', float), ('rho_sup', float), ('rho_vol', float)])

densities: dict[str, DensityTuple] = {"Al": [0.5, 1, 2],"Fe": [3, 4, 5], "Pb": [15, 20, 30]}
radii: list[float] = [1, 2, 3]

weight_calculation_methods: dict[str, Callable] = {
    "circle": lambda r, rho: 2 * pi * r * rho.rho_lin,
    "disk": lambda r, rho: pi * r**2 * rho.rho_sup,
    "spehere": lambda r, rho: 3/4 * pi * r**3 * rho.rho_vol 
}

for material_name, density_tuple in densities.items():
    for weight_type, weight_method in weight_calculation_methods.items():
        for radius in radii:
            print(f"{material_name} - {weight_type}: {weight_method(radius, density_tuple)}")
````
