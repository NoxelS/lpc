2\. Convert the following function into a pure function with no global variables or side effects

````python
x = 5
def f(alist):
    for i in range(x):
         alist.append(i)
    return alist

alist = [1,2,3]
ans = f(alist)
print (ans)
print (alist) # alist has been changed!

from collections.abc import Iterable

def f_pure(list: Iterable, *, x_range: int = 5) -> list:
     return [*list, *range(x_range)]

alist = [1,2,3]
ans = f_pure(alist)
print (ans)
print (alist) # alist has not been changed!

````

3\. Write a `decorator` hello that makes every wrapped function print “Hello!”, i.e. something like:

```python
@hello
def square(x):
    return x*x
```



````python
from collections.abc import Callable
from functools import wraps
from typing import Any

def hello(target_function: Callable):
    @wraps(target_function)
    def wrapper(*args, **kwargs) -> Any:
        print("Hello!")
        return target_function(*args, **kwargs)
    
    return wrapper

@hello
def square(x: float) -> float:
    return x*x

square(4)
````

4\. Write the factorial function so that it a) does and b) does not use recursion.

````python
from functools import reduce, lru_cache


@lru_cache(maxsize=256)
def fac_rec(n: int) -> int:
    if n < 0: raise ValueError("Invalid fac number")
    if n == 1 or n == 0: return 1
    return n * fac_rec(n-1)


def fac_non_rec(n: int) -> int:
    return reduce(lambda x,y: x * y, [m for m in range(1, n + 1)])

for i in range(1,6):
    print(f"{i} - rec: {fac_rec(i)} - nonrec: {fac_non_rec(i)}")
````

5. Use HOFs (zip in particular) to compute the weight of a circle, a disk and a sphere, assuming different radii and different densities:

```python
densities = {"Al":[0.5,1,2],"Fe":[3,4,5],"Pb": [15,20,30]}
radii = [1,2,3]
```

where the entries of the dictionary's values are the linear, superficial and volumetric densities of the materials respectively.

In particular define a list of three lambda functions using a comprehension that computes the circumference, the area and the volume for a given radius.


````python

import numpy as np
from math import gamma

densities: dict[str, list] = {"Al": [0.5, 1, 2],"Fe": [3, 4, 5], "Pb": [15, 20, 30]}
radii: list[float] = [1, 2, 3]

# Volume functions of a d-dimensional ball
V = [(lambda R, d=d: np.pi**(d/2) * np.pow(R, d) / gamma(d/2 + 1)) for d in range(1, 4)]

for material_name, density_tuple in densities.items():
    for rho, r, v, d in zip(density_tuple, radii, V, range(1, 4)):
        print(f"m_{material_name}^({d})({rho=}, {r=}) = {v(r) * rho}")

# One-Liner Solution
V = [(lambda R, d=d: np.pi**(d/2) * np.pow(R, d) / gamma(d/2 + 1)) for d in range(1, 4)]
print(*[(f"m_{material_name}^({d})({rho=}, {r=}) = {v(r) * rho}") for rho, r, v, d in zip(density_tuple, radii, V, range(1, 4)) for material_name, density_tuple in densities.items()], sep="\n")
````

````python

class Dog:

    species = 'mammal'

    def __init__(self, name: str, age: int, is_hungry = True):
        self.name = name
        self.age = age
        self.is_hungry = is_hungry

    def description(self) -> str:
        return f"'{self.name}' is {self.age} years old" # Use f-strings instead of format for static templates

    # Even better 
    def __str__(self): return self.description()

    def speak(self, sound: str) -> str:
        return f"'{self.name}' says {sound}"

    def eat(self) -> None:
        self.is_hungry = False

class RusselTerrier(Dog):
    def run(self, speed: str):
        return f"'{self.name}' runs {speed}"

class Bulldog(Dog):
    def run(self, speed: str):
        return f"'{self.name}' runs {speed}"


my_dogs = [
    *[RusselTerrier(f"RT-{num}", num, num % 2) for num in range(1, 2)],
    *[Bulldog(f"B-{num}", num, num % 2) for num in range(1, 2)]
]

# Feed all dogs
for dog in my_dogs: dog.eat()

# Temporary variables
all_mammels = np.all([dog.species == 'mammel' for dog in my_dogs])
all_hungry = np.all([dog.is_hungry for dog in my_dogs])
any_hungry = np.all([dog.is_hungry for dog in my_dogs])

print(f"I have {len(my_dogs)} dogs", end=". ")
print(*my_dogs, sep=". ", end=". ")

if any_hungry and all_hungry: print("My dogs are all hungry. ", end="")
if any_hungry and not all_hungry: print("Some of my dogs are hungry. ", end="")
if not any_hungry: print("None of my dogs is hungry", end="")
````
